```python
import pandas as pd
import calendar

# Hour
df['Hour'] = df['Time'].apply(lambda x: x.hour)

# Period
def get_period(hour):
    if 8 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 16:
        return 'Afternoon'
    elif 16 <= hour < 24:
        return 'Evening'
    else:
        return 'Night'

df['Period'] = df['Hour'].apply(get_period)

# Month Name
df['Month'] = df['Date'].dt.month
df['Month'] = df['Month'].apply(lambda x: calendar.month_name[x])
```

```python
X = df[['Sales', 'Quantity', 'Unit price']]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
cluster_summary = df.groupby('Cluster')[['Sales','Quantity','Unit price']].mean()
print(cluster_summary)
```

```python
def label_cluster(row):
    if row['Cluster'] == 0:
        return 'Low spenders'
    elif row['Cluster'] == 1:
        return 'Regular customers'
    else:
        return 'High value customers'

df['Customer_Type'] = df.apply(label_cluster, axis=1)
```

### Market Basket Analysis for Product Recommendations

To identify which product categories are often purchased together, we'll perform Market Basket Analysis. This involves:
1. Creating a one-hot encoded dataset of product lines for each invoice.
2. Using the Apriori algorithm to find frequent itemsets (combinations of product lines).
3. Generating association rules from these itemsets to recommend products.

```python
from mlxtend.frequent_patterns import apriori, association_rules

# Prepare data for Market Basket Analysis
# Group by Invoice ID and Product line, then create a 'basket' format
basket = (df.groupby(['Invoice ID', 'Product line'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('Invoice ID'))

# Convert quantities to a binary format (1 if product was bought, 0 otherwise)
def encode_units(x):
    if x <= 0:
        return 0
    if x >= 1:
        return 1

basket_sets = basket.applymap(encode_units)

display(basket_sets.head())
```

```python
# Find frequent itemsets using the Apriori algorithm
frequent_itemsets = apriori(basket_sets, min_support=0.001, use_colnames=True)

display(frequent_itemsets.head())
```

```python
# Generate association rules
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)

# Sort rules by confidence and lift
rules = rules.sort_values(['confidence', 'lift'], ascending=[False, False])

display(rules.head())
```

The generated association rules can now be used for recommendations. For a new customer who buys a product (antecedent), we can recommend the consequent product line. If the user wants to filter these rules by gender, we would need to run this analysis separately for male and female customer transactions, or incorporate gender into the itemsets if we want rules like 'if female buys X, then Y'.

```python
def recommend_by_gender(gender, top_n=3):
    """
    Recommends top product lines based on gender.

    Args:
        gender (str): The gender of the customer ('Male' or 'Female').
        top_n (int): The number of top product lines to recommend.

    Returns:
        pandas.Series: A Series of recommended product lines and their total sales.
    """
    if gender.lower() == 'female':
        # Use female_sales directly
        recommendations = female_sales.nlargest(top_n)
    elif gender.lower() == 'male':
        # Use male_sales directly
        recommendations = male_sales.nlargest(top_n)
    else:
        return "Invalid gender. Please specify 'Male' or 'Female'."

    print(f"Top {top_n} product recommendations for {gender} customers:")
    return recommendations
```

### Demo of the Recommendation Function

```python
# Example usage:
display(recommend_by_gender('Female'))
display(recommend_by_gender('Male', top_n=2))
```

```python
def recommend_by_gender_and_month(gender, month, top_n=2):
    """
    Recommends top product lines based on gender and a specific month.

    Args:
        gender (str): The gender of the customer ('Male' or 'Female').
        month (str): The month for which to get recommendations (e.g., 'March').
        top_n (int): The number of top product lines to recommend.

    Returns:
        pandas.Series: A Series of recommended product lines and their total sales.
    """
    filtered_data = df[df['Month'].str.lower() == month.lower()]

    if gender.lower() == 'female':
        gender_data = filtered_data[filtered_data['Gender'].str.lower() == 'female']
        sales_by_product = gender_data.groupby('Product line')['Sales'].sum()
        recommendations = sales_by_product.nlargest(top_n)
    elif gender.lower() == 'male':
        gender_data = filtered_data[filtered_data['Gender'].str.lower() == 'male']
        sales_by_product = gender_data.groupby('Product line')['Sales'].sum()
        recommendations = sales_by_product.nlargest(top_n)
    else:
        return "Invalid gender. Please specify 'Male' or 'Female'."

    if recommendations.empty:
        return f"No sales data found for {gender} customers in {month}."

    print(f"Top {top_n} product recommendations for {gender} customers in {month}:")
    return recommendations
```

### Demo of Monthly Gender-Based Recommendation Function

```python
# Example usage for March:
display(recommend_by_gender_and_month('Male', 'March', top_n=2))
display(recommend_by_gender_and_month('Female', 'March', top_n=2))
```
